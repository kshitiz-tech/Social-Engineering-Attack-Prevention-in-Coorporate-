```python
import pandas as pd
def analyze_column(df, column_name, start_day, end_day):
    filtered = df[(df['day'] >= start_day) & (df['day'] <= end_day)]
    col = filtered[column_name]
    
    print(f"\nFiltered Data ({column_name}, Days {start_day}–{end_day}):")
    print(filtered[['day', column_name]])
    
    print("\nSummary Statistics:")
    print(col.describe())
    
    print("\nTotal sum:", col.sum())
    print("Mean:", col.mean())


# Load file once
df = pd.read_csv("priebe_scan_result_pivot.csv")
df['day'] = pd.to_numeric(df['day'], errors='coerce')

# Run analyses
analyze_column(df, '6600', 200, 211)
analyze_column(df, '6601', 215, 226)
analyze_column(df, '6602', 300, 304)

```

```python
def analyze_specific_days(df, column_name, days_list):
    filtered = df[df['day'].isin(days_list)]
    col = filtered[column_name]

    print(f"\nFiltered Data ({column_name}):")
    print(filtered[['day', column_name]])

    print("\nSummary Statistics:")
    print(col.describe())

    print("\nTotal sum:", col.sum())
    print("Mean:", col.mean())


# Usage
specific_days = [50, 100, 120, 190, 210, 280, 290, 360, 380, 410]
specific_days2 = [458, 459, 460, 462, 465, 467, 472, 476, 495, 521, 559, 626]

analyze_specific_days(df, '6603', specific_days)
analyze_specific_days(df, '1023', specific_days2)

```
