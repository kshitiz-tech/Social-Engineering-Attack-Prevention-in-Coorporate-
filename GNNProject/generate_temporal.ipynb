```python
import pandas as pd
import re

print("Loading emails.csv...")
# Read the dataset. Enron emails can sometimes have weird characters, 
# so we'll ensure we drop any completely broken lines (though standard loads usually work fine).
df = pd.read_csv('emails.csv')

print("Extracting years from the message headers...")
# Use regex to find "Date:", skip some characters, and grab the first 4-digit number.
# flags=re.IGNORECASE handles cases where it might be "date:" instead of "Date:"
df['year'] = df['message'].str.extract(r'Date:.*?\b(\d{4})\b', flags=re.IGNORECASE)

# Convert the extracted year to a numeric type so we can filter easily
df['year'] = pd.to_numeric(df['year'], errors='coerce')

# Filter for Training set (1999, 2000)
print("Filtering Training Data (1999-2000)...")
train_df = df[df['year'].isin([1999, 2000])].copy()

# Filter for Testing set (2001, 2002)
print("Filtering Testing Data (2001-2002)...")
test_df = df[df['year'].isin([2001, 2002])].copy()

# Drop the temporary 'year' column so your output files have the exact same format as the input
train_df = train_df.drop(columns=['year'])
test_df = test_df.drop(columns=['year'])

print(f"Found {len(train_df)} emails for Training.")
print(f"Found {len(test_df)} emails for Testing.")

# Save to new CSV files
print("Saving to CSV...")
train_df.to_csv('train_emails_99_00.csv', index=False)
test_df.to_csv('test_emails_01_02.csv', index=False)

print("Done! Files saved as train_emails_99_00.csv and test_emails_01_02.csv")
```
